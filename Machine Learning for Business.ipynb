- 2.6 Put all previous steps into functions, perform steps 2.1 to 2.5 on files 'geo data 1 csv' and 'geo data 2 csv'.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error


def processar_regiao(caminho_csv, nome_regiao):
    df = pd.read_csv(caminho_csv)
    X = df[['f0', 'f1', 'f2']]
    y = df['product']
    
    X_train, X_valid, y_train, y_valid = train_test_split(
        X, y, test_size=0.25, random_state=42
    )
    
    model = LinearRegression()
    model.fit(X_train, y_train)
    
    preds = model.predict(X_valid)
    
    mean_pred = preds.mean()
    reqm = mean_squared_error(y_valid, preds, squared=False)
    
    print(f"\n=== {nome_regiao} ===")
    print("Volume médio previsto:", mean_pred)
    print("REQM:", reqm)
    
    return preds, y_valid
```

```python
preds0, y0_valid = processar_regiao('geo_data_0.csv', 'REGIÃO 0')
preds1, y1_valid = processar_regiao('geo_data_1.csv', 'REGIÃO 1')
preds2, y2_valid = processar_regiao('geo_data_2.csv', 'REGIÃO 2')
```

- Region 0

Relatively high average predicted volume (~92).

High RMSE (~38), indicating large dispersion between actual and predicted values.

The model has limited accuracy in this region.

- Region 1

Lowest predicted average volume (~69).

Extremely low REQM (~0 89).

This indicates that linear regression describes the data in this region very well, with high precision.

Suggests a perfect almost linear relationship between features and booking volume.

- Region 2

### 4. Write a function to calculate profit from a set of selected oil wells and model predictions:

```python

budget_total = 100_000_000          
n_pocos = 200
preco_por_unidade = 4500            
investimento_por_regiao = budget_total
```

```python
import numpy as np

def calcular_lucro(predicoes, valores_reais):
    
    top_200 = np.argsort(predicoes)[-200:]
    
    volume_total = valores_reais.iloc[top_200].sum()
    
    receita = volume_total * preco_por_unidade
    lucro = receita - investimento_por_regiao
    
    return lucro, volume_total
```

```python
lucro_0, volume_0 = calcular_lucro(preds0, y0_valid)
lucro_1, volume_1 = calcular_lucro(preds1, y1_valid)
lucro_2, volume_2 = calcular_lucro(preds2, y2_valid)

print("REGIÃO 0 - Lucro:", lucro_0)
print("REGIÃO 1 - Lucro:", lucro_1)
print("REGIÃO 2 - Lucro:", lucro_2)
```

- Region 0 had the highest potential profit, significantly outperforming regions 1 and 2.
- Although Region 1 had the lowest model error (REQM), its average booking volume is lower, which limits total profit.
- Region 2 showed intermediate profit, but with high prediction error, which indicates greater uncertainty.

- Based on gross profit potential alone, Region 0 would be the preliminary choice for developing new oil wells.

### 5. Calculate risks and profit for each region:

```python
import numpy as np

def bootstrap_profit(preds, target, n_samples=1000, n_wells=200,
                     revenue_per_unit=4500, investment=100_000_000):
    profits = []

    for _ in range(n_samples):
        
        sample = preds.sample(n=len(preds), replace=True, random_state=None)
        
        top_200 = sample.sort_values(ascending=False).head(n_wells)
        
        real_volume = target.loc[top_200.index].sum()
        
        profit = real_volume * revenue_per_unit - investment
        profits.append(profit)

    profits = np.array(profits)

    mean_profit = profits.mean()
    ci_lower, ci_upper = np.percentile(profits, [2.5, 97.5])
    risk = (profits < 0).mean() * 100  

    return mean_profit, ci_lower, ci_upper, risk
    
```

```python
preds0 = pd.Series(preds0, index=y0_valid.index)
preds1 = pd.Series(preds1, index=y1_valid.index)
preds2 = pd.Series(preds2, index=y2_valid.index)
```

```python
mean_0, low_0, high_0, risk_0 = bootstrap_profit(preds0, y0_valid)
mean_1, low_1, high_1, risk_1 = bootstrap_profit(preds1, y1_valid)
mean_2, low_2, high_2, risk_2 = bootstrap_profit(preds2, y2_valid)


print(f"REGIÃO 0:")
print(f"Lucro médio: {mean_0:.2f}")
print(f"IC 95%: [{low_0:.2f}, {high_0:.2f}]")
print(f"Risco de prejuízo: {risk_0:.2f}%\n")

print(f"REGIÃO 1:")
print(f"Lucro médio: {mean_1:.2f}")
print(f"IC 95%: [{low_1:.2f}, {high_1:.2f}]")
print(f"Risco de prejuízo: {risk_1:.2f}%\n")

print(f"REGIÃO 2:")
print(f"Lucro médio: {mean_2:.2f}")
print(f"IC 95%: [{low_2:.2f}, {high_2:.2f}]")
print(f"Risco de prejuízo: {risk_2:.2f}%")
```

- After applying the bootstrapping technique with 1,000 samples, all regions presented a risk of loss of less than 2.5%. Among them, Region 0 demonstrated the highest expected average profit, in addition to a completely positive confidence interval.
- Therefore, Region 0 is the most suitable for the development of new oil wells. This choice is consistent with the previous analysis of potential profit and presents the best balance between expected return and risk.
